```python
import pandas as pd 
import numpy as np 
```

```python
flights = pd.read_csv('data/flights.csv')
flights.head(10)
```

```python
# travelCode and userCode are identifier for a complete unique travel . And userCode is identifier for a unique user . 
# we will drop travelCode , it is not needed to model creation .

flights.drop(['travelCode'],axis = 1 , inplace = True)
```

```python
# dropping usercode 
flights.drop(['userCode'],axis = 1 , inplace = True)
```

```python
flights.head()
```

```python
# one hot encoding the "from" and "to" column 

from sklearn.preprocessing import OneHotEncoder 

from_to_encoder = OneHotEncoder(sparse_output=False)

cols = ['from' , 'to']

encoded_arr = from_to_encoder.fit_transform(flights[cols])

encoded_arr
```

```python
from_to_encoder.categories_
```

```python
encoded_arr.shape
```

```python
flights['flightType'].unique()
```

```python
# There is ordinal order in flightType so we will do ordinal encoding

from sklearn.preprocessing import OrdinalEncoder
flightType_encoder = OrdinalEncoder(categories=[['firstClass' ,'premium' , 'economic']])

encoded_flightType = flightType_encoder.fit_transform(flights[['flightType']])
```

```python
# Scaling the encoded_flightType
from sklearn.preprocessing import StandardScaler 
flightType_scaler = StandardScaler()
encoded_flightType_scaled = flightType_scaler.fit_transform(encoded_flightType)
```

```python
flights.head()
```

```python
target = 'price'
```

```python
# processing time and distance features
from sklearn.preprocessing import PowerTransformer

cols = ['time' , 'distance']
time_dis_pt = PowerTransformer()

time_distance_transformed = time_dis_pt.fit_transform(flights[cols])
time_distance_transformed
```

```python
# processing agency 


agency_ohe = OneHotEncoder(sparse_output=False)
agency_encoded = agency_ohe.fit_transform(flights[['agency']])
agency_encoded
```

```python
agency_ohe.categories_
```

```python
# # Processing date 
# # converting date feature to datetime
# flights['date'] = pd.to_datetime(flights['date'])

# # creating year , month , day_of_week , day_of_month 
# flights['year'] = flights['date'].dt.year
# flights['month'] = flights['date'].dt.month
# flights['day_of_week'] = flights['date'].dt.dayofweek # Monday = 0 , sunday = 6
# flights['day_of_month'] = flights['date'].dt.day
```

```python
# Processing date

flights["date"] = pd.to_datetime(flights["date"])

# Basic date features
flights["year"] = flights["date"].dt.year
flights["month"] = flights["date"].dt.month
flights["day_of_week"] = flights["date"].dt.dayofweek
flights["day_of_month"] = flights["date"].dt.day

# New engineered features

# Quarter
flights["quarter"] = flights["date"].dt.quarter

# Weekend flag
flights["is_weekend"] = (
    flights["day_of_week"].isin([5, 6])
).astype(int)

# Cyclical month encoding
flights["month_sin"] = np.sin(
    2 * np.pi * flights["month"] / 12
)

flights["month_cos"] = np.cos(
    2 * np.pi * flights["month"] / 12
)

# Cyclical day of week encoding
flights["dow_sin"] = np.sin(
    2 * np.pi * flights["day_of_week"] / 7
)

flights["dow_cos"] = np.cos(
    2 * np.pi * flights["day_of_week"] / 7
)

flights.head()
```

```python
flights.head()
```

```python
# scaling year , month , day_of_week , day_of_month 
date_cols = [
    "year",
    "month",
    "day_of_week",
    "day_of_month",
    "quarter",
    "is_weekend",
    "month_sin",
    "month_cos",
    "dow_sin",
    "dow_cos"
]

date_scaler = StandardScaler()

date_features_scaled = date_scaler.fit_transform(
    flights[date_cols]
)
date_features_scaled
```

```python
final_matrix = np.concatenate([encoded_arr , encoded_flightType_scaled , time_distance_transformed , agency_encoded , date_features_scaled] , axis = 1)
final_matrix.shape 
```

```python
# saving all the encoders for future use using pickle : save in artifacts folder
#  save from_to_encoder , flightType_encoder , flightType_scaler , time_dis_pt ,agency_ohe , date_scaler , then save a list of date columns which needs to be first created from user input and then needs to be scaled using date_scaler : cols
import pickle 
objects_to_save = {
    "from_to_encoder.pkl" : from_to_encoder,
    "flightType_encoder.pkl": flightType_encoder,
    "flightType_scaler.pkl" : flightType_scaler,
    "time_dis_pt.pkl" : time_dis_pt,
    "agency_ohe.pkl" : agency_ohe,
    "date_scaler.pkl" : date_scaler , 
    "date_cols.pkl" : date_cols
}

for filename , obj in objects_to_save.items():
    with open(f"artifacts/{filename}" , "wb") as f:
        pickle.dump(obj , f)

print("All the encoders and scalers saved successfully")
```

Saving the Feature Names

```python
# Feature names for from and to OHE
from_to_feature_names = from_to_encoder.get_feature_names_out(
    ['from', 'to']
).tolist()

# Flight type
flight_type_feature_names = ["flightType_scaled"]

# Time and distance
time_distance_feature_names = ["time_transformed",
                               "distance_transformed"]

# Agency OHE
agency_feature_names = agency_ohe.get_feature_names_out(
    ['agency']
).tolist()

# Date features
date_feature_names = [
    f"{col}_scaled" for col in date_cols
]

# Final feature names
feature_names = (
    from_to_feature_names +
    flight_type_feature_names +
    time_distance_feature_names +
    agency_feature_names +
    date_feature_names
)

print(len(feature_names))

# save the feature names 
with open('artifacts/feature_names.pkl' , 'wb') as f:
    pickle.dump(feature_names , f)
```

```python
# getting unique distances between each unique combination of from and to cities


distance_map = (
    flights
    .groupby(["from", "to"])["distance"]
    .unique()
    .reset_index()
)

print(distance_map.head(10))
```

We can see that the distance between each unique from and to location is different. But also a single unique distance for for each unique pair of for and to cities.

Basically , we can directly infer the distance between from and to cities.We will save this mapping as a dictionary for later use in our Flask API.

```python
city_pair_distance = {}

for _,row in distance_map.iterrows():
    from_city = row["from"]
    to_city = row["to"]
    distance = row["distance"][0]
    #forward mapping
    city_pair_distance[(from_city , to_city)] = distance
    #reverse mapping
    city_pair_distance[(to_city , from_city)] = distance


with open("artifacts/city_pair_distance.pkl" , "wb") as f:
    pickle.dump(city_pair_distance , f)
```

```python
# Now inspecting for time feature

# Checking how many unique time values are there for each unique from and to city pair

time_variation = (
    flights
    .groupby(["from", "to"])["time"]
    .nunique()
    .reset_index(name="unique_time_count")
)
time_variation.sort_values("unique_time_count", ascending=False)
```

For time also , there is a single unique time for each unique pair of from and to cities. We will save this mapping as a dictionary for later use in our Flask API.



```python
route_time_df = (
    flights
    .groupby(["from", "to"])["time"]
    .agg(lambda x: x.mode().iloc[0])
    .reset_index()
)

city_pair_time = {}

for _, row in route_time_df.iterrows():
    from_city = row["from"]
    to_city = row["to"]
    time_val = row["time"]

    # forward
    city_pair_time[(from_city, to_city)] = time_val
    # reverse
    city_pair_time[(to_city, from_city)] = time_val


with open("artifacts/city_pair_time.pkl", "wb") as f:
    pickle.dump(city_pair_time, f)
```
